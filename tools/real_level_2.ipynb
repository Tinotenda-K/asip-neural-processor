```python
#documentation: useful for understanding
#youtube: https://www.youtube.com/watch?v=AYuyN8vvkAM
#youtube: https://www.youtube.com/watch?v=kcVfAq-Ocy8
#youtube: https://www.youtube.com/watch?v=0idoEomDc9E


import numpy as np  #numerical arrays / matrix ops
import keras       #high level neural networks API from what I understand
from keras.datasets import mnist #dataset of handwritten digits
from keras.layers import Dense, Flatten, Input #types of layers im going to use
from keras.models import Sequential #stack of layers, mostlikely liner in our case to make it easier
```

```python
# Loading the dataset and perform splitting
(x_train, y_train), (x_test, y_test) = mnist.load_data() #loiding mnist and crating tuples "train and test" that will be used to store multiple items in a single variable

print("x_train shape:", x_train.shape) #print shape of the training data to check if it's fine
print("y_train shape:", y_train.shape)

print("x_test shape:", x_test.shape)
print("y_test shape:", y_test.shape)


```

```python
# Peforming reshaping operation
x_train = x_train.reshape(x_train.shape[0], 28, 28, 1) #reshaping to 4D tensor (number of samples, height, width, channels), to keep it compatible with keras otherwise we will get an error
x_test = x_test.reshape(x_test.shape[0], 28, 28, 1)

print("x_train shape:", x_train.shape)  #not useful but just here to check if it's fine  (N, 28, 28, 1)
print("x_test shape:", x_test.shape)
```

```python
# Normalize pixel values between 0 and 1
x_train = x_train.astype("float32") / 255.0    #convert a pixel value to float 32 and then scale it to be between 0 and 1
x_test = x_test.astype("float32") / 255.0

```

```python
# Reshape the labels and encode them categorically to match with 10 outputs

y_train = keras.utils.to_categorical(y_train.flatten(), num_classes=10) #flattening to make sure it's 1D array and convert each integer label into a 10-dimensional binary vector
y_test = keras.utils.to_categorical(y_test.flatten(), num_classes=10)   #we do this because we have 10 output neurons so things still make sense
```

```python
class NeuralNetwork:
    def __init__(self, size, hidden_layer_size1, hidden_layer_size2, output_size):
        pass
    def forward_pass(self):
        pass
    def backward_pass(self):
        pass
    def compute_accuracy(self, x=None, y=None):
        print("compute_accuracy() placeholder")
    def train(self, x=None, y=None, epochs=None, batch_size=None):
        print("train() placeholder")

```

```python
# those are the placeholder object
dnn = NeuralNetwork(size=784, hidden_layer_size1=10,hidden_layer_size2=10, output_size=10)   #128 might be the best value after playing arround for a bit
dnn.train(x_train, y_train, epochs=18, batch_size=32)
dnn.compute_accuracy(x_test, y_test)
```

```python
# Building the model
model = Sequential()   # start stacking layers
model.add(Input(shape=(28, 28, 1)))      #specify the expected input shape
model.add(Flatten())  #flatten the 28x28 image into a 784 dimension vector
model.add(Dense(10, activation='relu'))  #relu is used
model.add(Dense(10, activation='relu'))                 #new line be careful
model.add(Dense(10, activation='softmax'))   #10 output neurons for 10 classes, softmax to get probabilities
model.summary()  #as said, it prints the summary of the model

```

```python
# Compiling the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])  #adam is an optimization algorithm, categorical_crossentropy is a loss function used for multi-class classification problems, accuracy is a metric to evaluate the model's performance

```

```python
# Training the model
# Make sure y_train has correct shape (batch_size,10)
model.fit(x_train, y_train, epochs=18, batch_size=32, validation_split=0.4) #20% of training data used for validation, we can increase it if we want more accuracy technically, but I don't see any differences
```

```python
#this is how to get the weight
for layer in model.layers:
    if isinstance(layer, keras.layers.Dense):   #The isinstance() function returns True if the specified object is of the specified type, otherwise False.   check if the layer is a Dense layer
        weights, biases = layer.get_weights()  #
        print(f"Layer: {layer.name}")
        print("Weights shape:", weights.shape)
        print("Biases shape:", biases.shape)
        break #lmao if u don't put this it will print the weights and biases of every layer, which is a lot of information, lost 30 min of my life to this
```

```python
# Printing model structure in detail

print("\nMODEL STRUCTURE")
total_layers = len([l for l in model.layers if isinstance(l, Dense)]) #isinstance is used to check if the layer is a dense layer
print(f"Total Dense layers: {total_layers}")

total_neurons = 0
total_weights = 0
total_biases = 0


#that's where the shit gets real
for i, layer in enumerate(model.layers):    #need to read about the enumerate class, for some reason it's not a loop
    if isinstance(layer, Dense):    #same func again
        weights, biases = layer.get_weights()
        n_inputs, n_outputs = weights.shape
        total_neurons += n_outputs
        total_weights += weights.size
        total_biases += biases.size

        #all the print functions
        print(f"\n--- Layer {i}: {layer.name} ---")
        print(f"Inputs: {n_inputs}, Outputs (neurons): {n_outputs}")
        print(f"Weights shape: {weights.shape}")
        print(f"Biases shape: {biases.shape}")
        print(f"Number of weights: {weights.size}")
        print(f"Number of biases: {biases.size}")

        print("\nWeights:")
        for j, w in enumerate(weights.flatten()):
            print(f"W{i}[{j}] = {w}")

        print("\nBiases:")
        for j, b in enumerate(biases.flatten()):
            print(f"b{i}[{j}] = {b}")
```

```python
#summury, not diffucult to understand tho
print("\nSUMMARY ")
print(f"Total Dense Layers: {total_layers}")
print(f"Total Neurons: {total_neurons}")
print(f"Total Weights: {total_weights}")
print(f"Total Biases: {total_biases}")
print(f"Total Parameters (Keras count): {model.count_params()}")

```

```python
# Evaluating the model
loss, accuracy = model.evaluate(x_test, y_test)
print(f'Test Loss: {loss}')
print(f'Test Accuracy: {accuracy}')
```

```python
dense_layers = [layer for layer in model.layers if isinstance(layer, keras.layers.Dense)]
layer_sizes = [layer.units for layer in dense_layers]
num_layers = len(dense_layers)
```

```python
#let's create the txt file
with open("level_2_weights.txt", "w") as f:
    # Line 1: number of layers
    f.write(f"{num_layers}\n")

    # Line 2: number of neurons in each layer
    f.write(" ".join(map(str, layer_sizes)) + "\n")

    # Lines 3–(2 + num_layers): weights
    for layer in dense_layers:
        weights, biases = layer.get_weights()
        weights_flat = weights.flatten()
        # ensure single-line write without internal newlines
        line = " ".join(map(lambda x: f"{x:.8f}", weights_flat))
        f.write(line.replace("\n", " ") + "\n")

    # Lines (3 + num_layers)–(2 + 2*num_layers): biases
    for layer in dense_layers:
        weights, biases = layer.get_weights()
        biases_flat = biases.flatten()
        line = " ".join(map(lambda x: f"{x:.8f}", biases_flat))
        f.write(line.replace("\n", " ") + "\n")

print("\nfile saved")
```
